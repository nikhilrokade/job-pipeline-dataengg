```python
import os

repo_root = (
    "/Workspace/Users/nikhilrokade9951@gmail.com/job_pipeline/job-pipeline-dataengg"
)


def print_tree(startpath):
    for root, dirs, files in os.walk(startpath):
        # Exclude hidden git internals from output
        dirs[:] = [d for d in dirs if not d.startswith(".git")]
        level = root.replace(startpath, "").count(os.sep)
        indent = " " * 4 * level
        print(f"{indent}{os.path.basename(root)}/")
        subindent = " " * 4 * (level + 1)
        for f in files:
            print(f"{subindent}{f}")


print_tree(repo_root)
```
